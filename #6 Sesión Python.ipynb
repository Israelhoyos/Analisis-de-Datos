Practica 29: Obtener un dataframe que consolide los datos de February + March 2019.csv. Crear una columna que me de 15 dias menos que la fecha Date y me la muestre en el formato dd-mm-yyyy, crear otra columna que nos ponga el mes de la columna Date en castellano.

```python
#Solución practica 29
#0.- Cargar las librerias
import pandas as pd
import locale
from datetime import timedelta as tiempo
#1.- Cargar February + March por individual
february_2019 = pd.read_csv(r"C:\Users\formacio\Downloads\February 2019 (1).csv")
march_2019 = pd.read_csv(r"C:\Users\formacio\Downloads\March 2019 (1).csv")
#2.- Concatenar los 2 dataframes
feb_mar = pd.concat([february_2019,march_2019],ignore_index=True)
#3.- Ver la info del df resultante
#feb_mar.info()
#4.- Voy a convertir el campo Date en datetime
feb_mar['Date'] = pd.to_datetime(feb_mar['Date'])
#5.- Voy a justar el locale para el idioma
locale.setlocale(locale.LC_TIME, 'es_ES.UTF-8')
#6.- Voy a decrementar 15 dias a la fecha en una nueva columna
feb_mar['Date_15'] = feb_mar['Date'] + tiempo(days=-15)
feb_mar['Date_15'] = feb_mar['Date_15'].dt.strftime('%d/%m/%Y')
#7.- Poner en una nueva columna el mes en letras en mayus
feb_mar['Mes'] = feb_mar['Date'].dt.strftime('%B').str.upper()
feb_mar
```

Practica 30: Partiendo del dataframe del trimestre, averiguar la tabla de frecuencias por CUSTOMER, de los clientes que son de Brazil, Panama o USA y cuyo nombre de cliente acabe por la letra n

```python
#Solución practica 30
#Voy a filtrar por Country con isin
filtro = primer_tri_2019['Country'].isin(['Brazil','USA','Panama'])
datos = primer_tri_2019[filtro]
#Voy a reasignar datos para solo los CUSTOMERS que terminan por n
filtro = datos['Customer'].str.endswith('n')
datos = datos[filtro]
#Con los datos filtrados, voy a hacer la tabla de frecuencias
datos['Customer'].value_counts()
```

```python
#Como convertimos una agrupación, cuyo indice deberia de ser una columna, en un dataframe, con reset_index() * convertir una serie en un dataframe
datos = datos['Customer'].value_counts().reset_index()
type(datos)
```

```python
#Ya tengo el DataFrame, lo vamos a guardar en un fichero csv. Ojo la ruta entera, incluye carpeta/subcarpeta/nombre fichero/extension y poner la r delante. Un parametro sep= con que caracter queremos separar los datos parametro index=False
datos.to_csv(r'C:\temp\frecuencia_clientes.csv',sep=';',index=False)
```

```python
#Guardar en un csv las frecuencias por Country del dataframe del trimestre, tanto las absolutas como las relativas
#Guardar en un dataframe las absolutas
absolutas = primer_tri_2019['Country'].value_counts().reset_index()
#Guardo en un dataframe las relativas
relativas = round(primer_tri_2019['Country'].value_counts(normalize=True) * 100, 2).reset_index()
#Combinar los 2 dataframes para tener uno solo, eso lo haremos con merge
total_frecuencias = pd.merge(absolutas,relativas,on="Country",how='inner')
total_frecuencias
```

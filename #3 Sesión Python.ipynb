El bucle infinito while (mientras), no tiene variable asociada, no tiene control de iteraciones, mientras se cumpla la condición el bucle se ejecuta, cuando ya no se cumple se para

```python
i = 1

while i<=5:
    print(i)
    i+=1
```

```python
#Bucle while True, se ejecuta de manera infinita y solo se puede detener si en el codigo, encuentra una instrucción break
i=0

while True:
    print("Linea")
    #Controlar incrementando la i en 1, en cada iteración
    i+=1
    if i>3:
        break
```

Practica 16: Solicitar los 3 numeros para pasarle a una función mediante el input, devolver con un texto, el orden correcto de los numeros, yo le paso el 22, 8, 6 y me tiene que devolver 6, 8, 22

Practica 17: Con un bucle while true imprimir los meses (Enero-Diciembre), hasta que encuentre el mes de Julio.

```python
#Practica 16 solución
def ordenar_3numeros(a=5,b=10,c=15):
    #Aplicar logica para aeriguar si el numero a va primero
    if a<=b and a<=c:       #Si se cumplen las 2 condiciones, significa que la a es el numero mas pequeño, va delante en el orden
        if b<=c:
            print("Numeros ordenados",a,b,c)
        else:
            print("Numeros ordenados",a,c,b)
    #Aplicar logica para averiguar si el numero b va primero
    elif b<=a and b<=c:    #Significa que el numero b va el primero
        if a<=c:
            print("Numeros ordenados",b,a,c)
        else:       #significa que la c es mas pequeño que la a
            print("Numeros ordenados",b,c,a)
    else:    #Significa que la c va el primero
        if a<=b:
            print("Numeros ordenados",c,a,b)
        else:
            print("Numeros ordenados",c,b,a)
```

```python
#Practica 17: Con un bucle while true imprimir los meses (Enero-Diciembre), hasta que encuentre el mes de Julio.

meses = ['Enero','Febrero','Marzo','Abril','Mayo','Junio','Julio','Agosto','Septiembre','Octubre','Noviembre','Diciembre']

while True:
    for mes in meses:
        print("El mes actual es", mes)
        if mes=="Julio":
            break
    #Poner break al nivel del for para romper el while True
    break
```

Las clases personalizadas en Python, tienen atributos, metodos especiales (solo sirven para crear o construir la clase, pero no para usarlos con una instancia), metodos personalizados. Una clase es como una plantilla que sirve para crear objetos. Una clase seria como los planos de un edificio
- Una clase se define con la palabra class
- El metodo especial para iniciar una clase seria init
- self = hace referencia a si mismo (es como un parametro que solo se puede ver dentro de la clase, pero es global dentro de la clase)

```python
#Vamos a crear la clase alumno, la clase alumno tendra 3 atributos (nombre, edad, estatura), tendra un metodo personalizado, saludo de bienvenida, restar_altura = para que le quite 10 cms a su altura, animos = para darte animos de cara al curso
class alumno:
    #Iniciamos con la clase init
    def __init__(self, nombre, edad, estatura):
        self.nombre = nombre
        self.edad = edad
        self.estatura = estatura
    #Crear un metodo personalizado
    def saludo(self):
        print("Hola",self.nombre,"te doy la bienvenida a este curso de Python 2" )
    #Metodo personalizado dar animos
    def animos(self):
        print("Estoy seguro",self.nombre,"que aprobaras este curso")
    #Metodo personalizado para restar 10 cms a la altura
    def restar_altura(self):
        print("Tu altura menos 10 cms es",self.estatura-0.10)
```

```python
#Vamos a crear una instancia de la clase alumno
alumno_1 = alumno("Pedro Gomez",57,1.74)
```

```python
#El metodo personalizado, lleva parentesis
alumno_1.saludo()
```

```python
#El atributo de una instancia de una clase no lleva parentesis
alumno_1.nombre
```

```python
alumno_1.edad
```

```python
marta_perez = alumno("Marta Perez Capdevila",38,1.67)
```

```python
marta_perez.animos()
```

```python
luis_marco = alumno("Luis Marco Gracia",27,1.74)
```

```python
luis_marco.animos()
luis_marco.restar_altura()
luis_marco.edad
```

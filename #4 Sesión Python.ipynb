Practica 18: Crearemos una clase que se llamara vehiculo y que contendra los siguientes atributos
- Marca
- Modelo
- Año
- Velocidad maxima
- Matricula
- Precio Vehiculo
- Etiqueta ECO por defecto sea la B
- Tasa ECO por defecto 5% del vehiculo
- Un metodo personalizado que se llamara ficha_vehiculo, nos ha de mostrar un texto, donde ver la marca, modelo, matricula y precio vehiculo
- Un metodo personalizado que averigue si el año del vehiculo es anterior al 2023, me devuelva un texto que pona "Su vehiculo matricula, debe pasar la ITV"
- Un metodo personalizado que nos calcule la velocidad optima del vehiculo
  = velocidad maxima menos un 20%
- Un metodo personalizado, que nos calcule el importe de la tasa a pagar, sobre el precio del vehiculo  
  Etiqueta A - tasa sera del 3,75%  
  Etiqueta B - Tasa sera del 5%  
  Etiqueta C - Tasa sera del 7,33%  

```python
#Solución practica 18
class vehiculo:
    def __init__(self, marca, modelo, año, velocidad, matricula, precio_vehiculo, etiqueta_eco = 'B', tasa_eco = 0.05):
        self.marca = marca
        self.modelo = modelo
        self.año = año
        self.velocidad = velocidad
        self.matricula = matricula
        self.precio_vehiculo = precio_vehiculo
        self.etiqueta_eco = etiqueta_eco
        self.tasa_eco = tasa_eco
    #Metodo personalizado
    def ficha_vehiculo(self):
        print(self.marca+'|'+self.modelo+'|'+self.matricula+'|'+str(self.precio_vehiculo))
    #Metodo para averiguar la ITV
    def itv(self):
        if self.año<2023:
            print("Este vehiculo matricula",self.matricula,"debe pasar la ITV")
        else:
            print("No tiene que pasar la ITV")
    def velocidad_optima(self):
        resultado = self.velocidad * (1 - 0.20)
        #base(200) * 1 = 200
        #base(200) * (1 + 0.20)
        #base(200) * (1 - 0.20)
        print("Su velocidad optima es",resultado,"Km por hora")
    def importe_tasa_eco(self):
        match self.etiqueta_eco:
            case "A":
                resultado = self.precio_vehiculo * 0.0375
            case "B":
                resultado = self.precio_vehiculo * 0.05
            case "C":
                resultado = self.precio_vehiculo * 0.0733

        print("El importe de la tasa eco para este vehiculo es",resultado)
            
```

Practica 19: Vamos a restar 2 fechas, la primera sera 11/04/2021 y la segunda 07/07/2025, mostrar en un texto cuantas meses han transcurrido entre las 2 fechas.

```python
#Solución practica 19
from datetime import date as dt
#Crear la instancia de la fecha1
fecha1 = dt(day=11,month=4,year=2021)
#crear la instancia de la fecha2
fecha2 = dt(day=7,month=7,year=2025)
#Comprobar los valores
print("la fecha1 es",fecha1,"y la fecha2 es",fecha2)
#Restar las 2 fechas, la mas nueva de la mas antigua
resultado = fecha2 - fecha1
meses = resultado.days / 365.25 * 12
#Mostrar el resultado
print("El numero de meses transcurridos entre las 2 fechas es de",int(meses))
```

```python
#la clase timedelta, sirve para avanzar / retroceder tiempo
from datetime import date, timedelta

hoy = date.today()

#Avanzar tiempo
resultado = hoy + timedelta(days=10)
print(resultado)

#Retroceder tiempo
resultado = hoy + timedelta(days=-10)
print(resultado.strftime('%d/%m/%Y'))
```

Practica 20: Averiguar que fecha sera dentro de 101 dias, mostrala en formato europeo y mostrar tambien el mes que sera en letras en mayusculas

Practica 21: Averiguar cada uno, cuantas horas llevais en este mundo, asumiendo que nacisteis a las 15:00:00

Practica 22: Averiguar cuantos minutos faltan para el dia 01/01/2028 a las 14:36:45

```python
#Solución practica 20
from datetime import date, timedelta

hoy = date.today()
fecha_101 = hoy + timedelta(days=101)

#Visualizar el resultado formateado
print("Dentro de 101 dias sera",fecha_101.strftime('%d/%m/%Y'))
print("El mes en letras dentro de 101 dias es",fecha_101.strftime('%B').upper())
```

```python
#Solución practica 21
from datetime import datetime as horas

nacimiento = horas(day=7,month=3,year=1963,hour=15,minute=0,second=0)
print("Mi fecha/hora de nacimiento es",nacimiento.strftime('%d/%m/%Y %H:%M:%S'))
ahora = horas.now()

#Calcular las horas restando la mas nueva, me dara segundos, como lo convierto a horas, dividir por 3600
resultado = ((ahora - nacimiento).total_seconds()) / 3600
#resultado ya son horas
print("El numero de horas que llevo en el planeta es",int(resultado))
```

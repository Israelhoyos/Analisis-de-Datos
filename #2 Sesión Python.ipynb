Practica 5: Mediante una función input, almacenaremos un numero entero y
lo clasificaremos por multiplos de 25, hasta el 150. Vigilar los numeros no validos (los validos seran del 0-150). Ejemplo: El 36 esta comprendido entre >25 y 50. Pero usando las funciones de logica and y or

```python
#Solución practica 5
valor = int(input("Escriba un numero entre 0 y 150"))
if valor<=0 or valor>150:
    print("Valor no valido, vuelva a intentarlo")
elif valor>0 and valor<=25:
    print("El valor esta comprendido entre >0-25")
elif valor>25 and valor<=50:
    print("El valor esta comprendido entre >25-50")
elif valor>50 and valor<=75:
    print("El valor esta comprendido entre >50-75")
```

Una estructura de decisión que se llama, match...case. Solo disponible a partir de la version 3.10 de python
Solo acepta valores iguales, tanto para numeros como para texto

```python
x = 34
match x:
    case 1:
        print("El valor es 1")
    case 10:
        print("El valor es 10")
    case 33:
        print("El valor es 33")
    case default:
        print("No se encuentra este valor")
```

Practica 6: mediante un input al que le pediremos un numero entero entre 1 y 12, mas una estructyra de decision, mostrar un texto que si el numero es 4 diga "El mes 4 es Abril", "El mes 8 es Agosto", etc...

```python
#Solución practica 6
mes = int(input("Indique el numero de mes, entre 1-12"))
if mes<1 or mes>12:
    print("Numero de mes, no valido")
else:
    match mes:
        case 1:
            print("El mes 1 es Enero")
        case 2:
            print("El mes 2 es Febrero")
        case 3:
            print("El mes 3 es Marzo")
        case default:
            print("Mes no valido")
```

```python
mes = int(input("Indique el numero de mes, entre 1-12"))
if mes<1 or mes>12:
    print("Numero de mes, no valido")
else:
    match mes:
        case 1:
            nombre_mes = "Enero"
        case 2:
            nombre_mes = "Febrero"
        case 3:
            nombre_mes = "Marzo"
        case default:
            print("Mes no valido")
            
print("El mes",mes,"es",nombre_mes)
```

Los bucles en Python; Un bucle es un codigo que se repite, hasta que se cumple una condición dada, los bucles pueden ser finitos o infinitos
Son finitos si sabemos de antemano el numero de veces que se ejecutara el coidgo del bucle, bucle for (Lleva asociada una variable)
Son infinitos si no sabemos cuantas veces se ejecutara (bucle while) (No lleva asociada ninguna variable)

```python
for i in [1,2,3,4,5]:
    print("El numero es",i)
```

```python
for i in [18,33,2,39,45,66,17]:
    print("El numero es",i)
```

```python
#Controlar el numero de iteración, creare la variable iter (variable contador)
iter = 0

for i in [18,33,2,39,45,66,17]:
    print("El numero es",i)
    #iter = iter + 1
    iter+=1

print("El numero de iteraciones, ha sido",iter)
```

Practica 9: Mediante un bucle for...Mostrar en el output de manera individual, los numeros de la siguiente lista [25,33,64,7,19,65,29,64,15]
siempre y cuando la suma acumulada de los numero de lista sea menor de 180

```python
#Solución practica 7
lista_meses = ['Enero','Febrero','Marzo','Abril','Mayo','Junio','Julio','Agosto','Septiembre','Octubre','Noviembre','Diciembre']

num_mes = 1

for mes in lista_meses:
    print("El numero de mes",num_mes,"es",mes)
    num_mes+=1
```

```python
#Solución practica 8
lista_meses = ['Enero','Febrero','Marzo','Abril','Mayo','Junio','Julio','Agosto','Septiembre','Octubre','Noviembre','Diciembre']
ordinal = ['primer','segundo','tercer','cuarto','quinto','sexto','septimo','octavo','noveno','decimo', 'onceavo', 'doceavo']

num_mes = 1

for mes in lista_meses:
    print("El",ordinal[num_mes-1],"mes del año es",mes)
    num_mes+=1
```

```python
#Solución practica 9
lista_numeros = [25,33,64,7,19,65,29,64,15]

acumulado = 0

for numero in lista_numeros:
    acumulado+=numero
    if acumulado<180:
        print("El numero es",numero,"y el acumulado es",acumulado)
```

Practica 12: Mediante un bucle, imprimir cada una de las letras de la palabra "Supercalifragilisticoespialidoso", menos las a y las e, ademas las imprimiremos en mayusculas, y si es una c imprimiremos un $

```python
#Solución practica 12
palabra = "Supercalifragilisticoespialidoso"

for letra in palabra:
    if letra=='a' or letra=='e':
        continue
    if letra=='c':
        print('$')
    else:
        print(letra.upper())
```

```python
#Serie numerica, para ver funciones de resumen
lista_numeros = [15,28,17,75,98,63,78,94,15,27,112,45]

serie_numeros = pd.Series(lista_numeros)
serie_numeros
```

```python
#Funciones de resumen sobre una serie numerica
print("La suma de los valores de la serie es",serie_numeros.sum())
print("El promedio de los valores de la serie es",serie_numeros.mean())
#Como redondear un numero decimal, con round
print("El promedio de los valores de la serie es",round(serie_numeros.mean(),2))
print("El numero mas grande de la serie es",serie_numeros.max())
print("El numero mas pequeño de la serie es",serie_numeros.min())
print("El numero de filas o registros es",serie_numeros.count())
```

```python
#Funciones de estadistica sobre una serie numerica
#"\n"
print("La suma acumulada es:"+"\n",(serie_numeros.cumsum()),sep="")
print()
#La varianza = mide cuanto se separan los datos respecto de su media aritmetica
print("La varianza es",round(serie_numeros.var()))
#La desviación tipica, es la raiz cuadrada de la varianza, devuelve las unidades originales
print("La desviación tipica es",round(serie_numeros.std(),2))
```

```python
#Resumen estadistico de una Serie
serie_numeros.describe()
```

```python
#Como transformar los datos de una serie numerica (operaciones vectorizadas)
serie_numeros * 4
#Es solo para el output, no modifica el original
```

```python
serie_numeros / 4
```

```python
#La ordenación de una serie, no modifica el original
serie_numeros.sort_values(ascending=False, ignore_index=True)
#Una copia ordenada y reindexada
#serie_numeros2 = serie_numeros.sort_values(ascending=True, ignore_index=True)
```

```python
#El filtrado de una serie, el filtrado va entre los corchetes
serie_numeros[serie_numeros<40].sort_values(ascending=True, ignore_index=True)
```

Practica 23: Partiendo de la lista VIP = [18,35,78,44,63,24,27,88,125,7,53,12,44,62,96], la convertiremos en un Serie de Pandas (serie_vip)
- Haremos una copia (VIPx3) que sera igual la serie_vip por 3
- Haremos una copia de la serie solo para los valores que sean menores de 70 (vip70)
- Ordenar la serie_vip en orden descendente y con los indices reindexados
- Ver en un texto, la suma de los valores de la serie_vip
- Ver en un texto la diferencia de la suma de serie_vip con la suma de la serie vip70
- Ver en un texto, el promedio redondeado a 3 decimales de la serie vip70
- Ver la desviación tipica de la serie_vip

```python
#Solución practica 23
import pandas as pd
#Crear la lista vip
lista_vip = [18,35,78,44,63,24,27,88,125,7,53,12,44,62,96]
#Convertir la lista en una serie
serie_vip = pd.Series(lista_vip)
#Haremos la copia vipx3
vipx3 = serie_vip * 3
#Una copia de la serie original, solamente para los valores menores de 70
vip70 = serie_vip[serie_vip<70]
#Ver la serie vip en orden descendente y reindexado
serie_vip.sort_values(ascending=False, ignore_index=True)
#Ver en un texto la suma de los valores de la serie vip
print("La suma de los valores de la serie vip es", serie_vip.sum())
#Ver en un texto la diferencia de la suma de serie_vip con la suma de la serie vip70
print("La diferencia de la suma de las 2 series es",serie_vip.sum()-vip70.sum())
#Ver en un texto, el promedio redondeado a 3 decimales de la serie vip70
print("El promedio de la serie vip70 es",round(vip70.mean(),3))
#Ver la desviación tipica de la serie_vip
print("La desviación tipica de la serie_vip", serie_vip.std())
```

```python
#Como crear un dataframe de manera manual
import pandas as pd

pedidos = pd.DataFrame([
    {"Cliente":"Comercial de Suministros","Cantidad":29,"Precio Unidad":4.75},
    {"Cliente":"Almacenes Sanchez","Cantidad":52,"Precio Unidad":6.19},
    {"Cliente":"Distribuciones Marquez","Cantidad":64,"Precio Unidad":11.60},
    {"Cliente":"Productos industriales","Cantidad":41,"Precio Unidad":2.90},
    {"Cliente":"Manufacturas Gomez","Cantidad":33,"Precio Unidad":9.51},
])
#LLamo al dataframe para verlo en formato dataframe
pedidos
#Pandas sin formato
print(pedidos)
```

```python
#Ver los primeros n registros del dataframe
pedidos.head(2)
```

```python
#Ver los ultimos n registros del dataframe
pedidos.tail(2)
```

```python
#Ver solamente una columna de un dataframe = voy a ver una serie
pedidos['Cliente']
```

```python
#Ver mas de una columna, +1 columna = dataframe
pedidos[['Cliente','Cantidad']]
```

```python
#Como añadir una columna en un dataframe
#Opción 1 - Poniendo un valor fijo para todas las filas
#Opción 2 - Usando una o varias columnas existentes
#Opción 3 - Con el metodo apply (usando una función generica o especifica)
#Opción 4 - Con los metodos loc e iloc, recorriendo el dataframe fila a fila
```

```python
#Si no encuentra el nombre de una columna la crea
pedidos['Empresa'] = "Mi Empresa 1"
pedidos
```

```python
#Voy a crear la columna importe (precio unidad por cantidad)
pedidos['Importe'] = round(pedidos['Precio Unidad'] * pedidos['Cantidad'],1)
pedidos
```

```python
#Vamos a crear una función y aplicarla a cada registro de un dataframe, con la función apply (metodo especifico)
def descuento(maria):
    if maria>50:
        return "Si"
    else:
        return "No"

#Aplicar la función para la nueva columna
pedidos['Descuento'] = pedidos['Cantidad'].apply(descuento)
pedidos
```

```python
#Metodo generico, que significa, ya no llamo a apply desde un campo del dataframe, si no desde todo el dataframe. Podras usar todos los campos
def calcula_descuento(fila):    #fila no es un valor, es una fila que contiene todos las columnas del dataframe
    if fila['Cantidad']>50 and fila['Precio Unidad']<10:
        return fila['Importe'] * (1-0.20)  #20% dto
    else:
        return fila['Importe']

#Aplicar el metodo generico, axis=1 va por columna, axis = 0 para filas
pedidos['Descuento2'] = pedidos.apply(calcula_descuento, axis=1)
pedidos
```

```python
#Añadir una columna con loc (loc trabaja con nombre de columna y posicion de fila)
pedidos.loc[:,'Diferencia Importe'] = [23,54,63,42,98]
pedidos
```

```python
#Añadir con iloc = integer location
pedidos.iloc[0,0]   #el primer 0 es el indice de la fila, el segundo el indice de la columna, podemos leer o podemos escrinir
```

```python
pedidos.loc[0,'Cliente']
```

```python
pedidos.iloc[0,0] = "Cliente 20"
```

Practica 24: Aprovechando el dataframe pedidos, crear la columna "Precio_Unidad_2", que incremente "Precio Unidad" en un 15%

Practica 25: Aprovechando el dataframe pedidos, crear la columna
"Cantidad_3", que incremente el campo "Cantidad" en un 25%, si la "Cantidad" es mayor de 45, si no poner la misma "Cantidad"

Practica 26: Aprovechando el dataframe pedidos, crear la columna "diferencia_cantidad", que me de la diferencia entre la cantidad mas grande de toda la serie y la cantidad de cada fila

```python
#Solución practica 24
pedidos['Precio_Unidad_2'] = round(pedidos['Precio Unidad'] * 1.15, 2)
pedidos
```

```python
#Solución practica 25, lo resolvemos con apply
def incremento_cantidad(cantidad):
    if cantidad>45:
        return int(cantidad * (1 + 0.25))
    else:
        return cantidad

#Llamar a la función con apply
pedidos['Cantidad_3'] = pedidos['Cantidad'].apply(incremento_cantidad)
pedidos
```

```python
#Solución practica 26
cant_max = pedidos['Cantidad'].max()

pedidos['diferencia_cantidad'] = cant_max - pedidos['Cantidad']
pedidos
```

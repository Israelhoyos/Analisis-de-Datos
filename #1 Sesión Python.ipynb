Practica 2: Crear una lista vacia llamada meses, despues le añadiremos los meses de enero a junio (menos abril), añadiremos al final el mes de Julio, insertaremos en su lugar el mes de abril, a continuación añadiremos de una vez los meses de Agosto a Diciembre, finalmente añadir al final 2 veces el mes de Junio.
- Guardar en una variable el numero de veces que esta Junio
- Eliminar los meses sobrantes
- Mostrar la lista final
- Mostrar en un texto el numero de veces que estaba repetido "Junio"

```python
#Solución Practica 2
meses = []
print(meses)
meses = ["Enero","Febrero","Marzo","Mayo","Junio"]
print(meses)
meses.append("Julio")
print(meses)
meses.insert(3,"Abril")
print(meses)
meses2 = ['Agosto','Septiembre','Octubre','Noviembre','Diciembre']
meses.extend(meses2)
print(meses)
meses.append('Junio')
meses.append('Junio')
print(meses)
#Guardar en una variable, el numero de veces de Junio
junio = meses.count("Junio")
#Eliminar los meses sobrantes
len(meses)
del meses[12:]
print(meses)
print('***********************')
print("El numero de veces que esta repetido Junio es:",junio)
```

Estructuras de decision en python, las estructuras siempre terminan con : (2 puntos) y las lineas de codigo que pertenecen a esa estructuras, van identadas cuatro espacios a la derecha
Operadores de comparación:
> mayor que
>= mayor o igual que
< menor que
<= menor o igual que
== igual que
!= distinto que
Un orperador de comparación, devuelve True o False

```python
x=3
if x>10:
    print(x,"si que es mayor de 10")
else:
    print(x, "No es mayor de 10")
print("If terminado")
```

```python
#Si necesito mas opciones de devolución, usaremos elif (debe volver a analizar la condicion)
x = 730
if x<=30:
    print(x, "esta comprendido entre 0-30")
elif x<=60:
    print(x, "esta comprendido entre >30-60")
elif x<=100:
    print(x, "esta comprendido entre >60-100")
else:
    print(x, "es mayor que 100")
```

Practica 3: Mediante una función input alamcenar un valor numerico, y le sumaremos 8.56 si ese valor es mayor de 50, si no le restaremos 14.27

Practica 4: Mediante una función input, almacenaremos un numero entero y
lo clasificaremos por multiplos de 25, hasta el 150. Vigilar los numeros no validos (los validos seran del 0-150). Ejemplo: El 36 esta comprendido entre >25 y 50

```python
#Solución practica 3
numero = float(input("Ponga el valor numerico"))
if numero>50:
    print("El resultado es",round(numero+8.56,2))
else:
    print("El resultado es",round(numero-14.27,2))
```

```python
#Solución practica 4
numero = int(input("Ponga el valor numerico"))
if numero<0:
    print("El valor no es valido")
elif numero<=25:
    print("El numero",numero,"esta comprendido entre 0-25")
elif numero<=50:
    print("El numero",numero,"esta comprendido entre >25-50")
elif numero<=75:
    print("El numero",numero,"esta comprendido entre >50-75")
elif numero<=100:
    print("El numero",numero,"esta comprendido entre >75-100")
elif numero<=125:
    print("El numero",numero,"esta comprendido entre >100-125")
elif numero<=150:
    print("El numero",numero,"esta comprendido entre >125-150")
else:
    print("El valor no es valido")
```
